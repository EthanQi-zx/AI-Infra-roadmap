```python
A = torch.arange(40, dtype=torch.float32).reshape(2, 5, 4)
A, A.sum(), A.sum(axis=0), A.sum(axis=1), A.sum(axis=2), A.sum(axis=[0, 1])
```

```python
# 求平均值
A = torch.arange(20, dtype=torch.float32).reshape(5, 4)
A.mean(), A.sum() / A.numel()
```

```python
# 求每一列的平均值
print(A)
A.mean(axis=0), A.sum(axis=0) / A.shape[0]
```

```python
# 求每一行的和，并保持不降维，只是把对应维度降为1
sum_A = A.sum(axis=1, keepdims=True)
sum_A, A.shape, sum_A.shape
```

```python
x = torch.ones(4, dtype=torch.float32)
y = torch.arange(4, dtype=torch.float32)
x, y, torch.dot(x, y)   # 点积是相同位置的元素乘积的和
```

```python
torch.sum(x * y)   # 另一种计算点积的方式
```

```python
A = torch.arange(20, dtype=torch.float32).reshape(5, 4)
x = torch.arange(4, dtype=torch.float32)
A, x, torch.mv(A, x)   # 矩阵-向量积
```

```python
B = torch.arange(12, dtype=torch.float32).reshape(4, 3)
torch.mm(A, B)   # 矩阵-矩阵积
```

```python
# 求向量的L2范数：每个元素平方和的平方根
u = torch.tensor([3.0, -4.0])
torch.norm(u)
```

```python
# 求向量的L1范数：每个元素绝对值的和
torch.abs(u).sum()
```
